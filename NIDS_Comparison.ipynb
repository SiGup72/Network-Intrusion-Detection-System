```python
import json
import pandas as pd

results_path = "/content/drive/MyDrive/NIDS/results"

# Load saved metrics
with open(f"{results_path}/logistic_regression_metrics.json", "r") as f:
    lr_results = json.load(f)

with open(f"{results_path}/random_forest_metrics.json", "r") as f:
    rf_results = json.load(f)

with open(f"{results_path}/xgboost_metrics.json", "r") as f:
    xgb_results = json.load(f)

# Combine results
results = [lr_results, rf_results, xgb_results]

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
comparison_display = comparison_df.copy()

comparison_display["training_time_minutes"] = (
    comparison_display["training_time_seconds"] / 60
).round(2)

for col in [
    "accuracy",
    "macro_precision",
    "macro_recall",
    "macro_f1",
    "weighted_f1"
]:
    comparison_display[col] = comparison_display[col].round(4)

comparison_display = comparison_display[
    [
        "model",
        "training_time_minutes",
        "accuracy",
        "macro_precision",
        "macro_recall",
        "macro_f1",
        "weighted_f1"
    ]
]

comparison_display
```
